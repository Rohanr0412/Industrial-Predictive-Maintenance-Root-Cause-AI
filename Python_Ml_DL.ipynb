```python
# Make sure failures are sorted
failures_sorted = failures.sort_values(["machineID", "datetime"]).copy()

# Keep original telemetry unchanged
telemetry_sorted = telemetry.sort_values(["machineID", "datetime"]).copy()

# Store sensor values observed in the 24 hours before each failure
pre_failure_records = []

for _, failure_row in failures_sorted.iterrows():

    machine = failure_row["machineID"]
    failure_time = failure_row["datetime"]

    start_time = failure_time - pd.Timedelta(hours=24)

    window = telemetry_sorted[
        (telemetry_sorted["machineID"] == machine) &
        (telemetry_sorted["datetime"] >= start_time) &
        (telemetry_sorted["datetime"] < failure_time)
    ]

    if len(window) > 0:
        window = window.copy()
        window["failure_time"] = failure_time
        window["failure_component"] = failure_row["failure"]
        pre_failure_records.append(window)

pre_failure = pd.concat(pre_failure_records, ignore_index=True)

print("Pre-failure telemetry shape:", pre_failure.shape)

display(pre_failure.head())
```

```python
sensor_columns = ["volt", "rotate", "pressure", "vibration"]

comparison = pd.DataFrame({
    "Overall_Mean": telemetry[sensor_columns].mean(),
    "Pre_Failure_Mean": pre_failure[sensor_columns].mean(),
    "Overall_Std": telemetry[sensor_columns].std(),
    "Pre_Failure_Std": pre_failure[sensor_columns].std()
})

comparison["Mean_Difference"] = (
    comparison["Pre_Failure_Mean"] -
    comparison["Overall_Mean"]
)

comparison["Mean_Difference_%"] = (
    comparison["Mean_Difference"] /
    comparison["Overall_Mean"] * 100
).round(2)

display(comparison)
```

```python
# Feature Engineering — Cell 1
# Create clean working dataset

df = telemetry.copy()

df["datetime"] = pd.to_datetime(df["datetime"])

df = df.sort_values(
    ["machineID", "datetime"]
).reset_index(drop=True)

print("Shape:", df.shape)
print("Date range:", df["datetime"].min(), "to", df["datetime"].max())
print("Machines:", df["machineID"].nunique())

df.head()
```

```python
# Feature Engineering — Cell 2
# Rolling sensor statistics

sensor_columns = ["volt", "rotate", "pressure", "vibration"]

for col in sensor_columns:

    # Rolling means
    df[f"{col}_mean_3h"] = (
        df.groupby("machineID")[col]
        .transform(lambda x: x.rolling(3, min_periods=1).mean())
    )

    df[f"{col}_mean_6h"] = (
        df.groupby("machineID")[col]
        .transform(lambda x: x.rolling(6, min_periods=1).mean())
    )

    df[f"{col}_mean_24h"] = (
        df.groupby("machineID")[col]
        .transform(lambda x: x.rolling(24, min_periods=1).mean())
    )

    # 24-hour variability
    df[f"{col}_std_24h"] = (
        df.groupby("machineID")[col]
        .transform(lambda x: x.rolling(24, min_periods=2).std())
    )

print("Shape:", df.shape)

df.head()
```

```python
# Feature Engineering — Cell 3
# Sensor change and trend features

for col in sensor_columns:

    # Change from previous hour
    df[f"{col}_change_1h"] = (
        df.groupby("machineID")[col]
        .diff(1)
    )

    # Change from 3 hours ago
    df[f"{col}_change_3h"] = (
        df.groupby("machineID")[col]
        .diff(3)
    )

    # Change from 24 hours ago
    df[f"{col}_change_24h"] = (
        df.groupby("machineID")[col]
        .diff(24)
    )

print("Shape:", df.shape)

df.head()
```

```python
# Feature Engineering — Cell 4
# Historical error features

errors_sorted = errors.copy()

errors_sorted["datetime"] = pd.to_datetime(errors_sorted["datetime"])

errors_sorted = errors_sorted.sort_values(
    ["machineID", "datetime"]
)

# Create binary error event
errors_sorted["error_event"] = 1

# Merge errors onto telemetry timeline
df = pd.merge_asof(
    df.sort_values("datetime"),
    errors_sorted[
        ["datetime", "machineID", "errorID", "error_event"]
    ].sort_values("datetime"),
    on="datetime",
    by="machineID",
    direction="backward",
    tolerance=pd.Timedelta(hours=24)
)

print("Shape:", df.shape)

df.head()
```

```python


# Remove temporary columns if they exist
df = df.drop(
    columns=["errorID", "error_event"],
    errors="ignore"
)

# Make sure datetime is datetime
errors["datetime"] = pd.to_datetime(errors["datetime"])

# Create an error event table
error_events = (
    errors[["machineID", "datetime"]]
    .assign(error_event=1)
    .drop_duplicates()
)

# Combine telemetry timestamps and error events
error_events = error_events.sort_values(
    ["machineID", "datetime"]
)

# Count errors in previous 24 hours
error_24h = (
    error_events
    .set_index("datetime")
    .groupby("machineID")["error_event"]
    .rolling("24h", closed="left")
    .sum()
    .reset_index()
    .rename(columns={"error_event": "error_count_24h"})
)

# Count errors in previous 7 days
error_7d = (
    error_events
    .set_index("datetime")
    .groupby("machineID")["error_event"]
    .rolling("7D", closed="left")
    .sum()
    .reset_index()
    .rename(columns={"error_event": "error_count_7d"})
)

print("24h error table:", error_24h.shape)
print("7d error table:", error_7d.shape)
```

```python
# Feature Engineering — Cell 4D
# Corrected merge of historical error counts

# Make sure datetime is datetime
df["datetime"] = pd.to_datetime(df["datetime"])
error_24h["datetime"] = pd.to_datetime(error_24h["datetime"])
error_7d["datetime"] = pd.to_datetime(error_7d["datetime"])

# IMPORTANT:
# merge_asof requires sorting primarily by the datetime column

df = df.sort_values(
    ["datetime", "machineID"]
).reset_index(drop=True)

error_24h = error_24h.sort_values(
    ["datetime", "machineID"]
).reset_index(drop=True)

error_7d = error_7d.sort_values(
    ["datetime", "machineID"]
).reset_index(drop=True)


# Merge 24-hour error count
df = pd.merge_asof(
    df,
    error_24h,
    on="datetime",
    by="machineID",
    direction="backward",
    allow_exact_matches=False
)


# Merge 7-day error count
df = pd.merge_asof(
    df,
    error_7d,
    on="datetime",
    by="machineID",
    direction="backward",
    allow_exact_matches=False
)


# No previous errors → 0
df["error_count_24h"] = df["error_count_24h"].fillna(0)
df["error_count_7d"] = df["error_count_7d"].fillna(0)


print("Shape:", df.shape)

df[[
    "datetime",
    "machineID",
    "error_count_24h",
    "error_count_7d"
]].head(20)
```

```python
# Feature Engineering — Cell 5
# Maintenance history features

# Prepare maintenance data
maintenance_history = maintenance.copy()

maintenance_history["datetime"] = pd.to_datetime(
    maintenance_history["datetime"]
)

maintenance_history = maintenance_history.sort_values(
    ["datetime", "machineID"]
).reset_index(drop=True)


# Create maintenance event
maintenance_history["maintenance_event"] = 1


# Count maintenance events in previous 30 days
maintenance_30d = (
    maintenance_history[
        ["machineID", "datetime", "maintenance_event"]
    ]
    .set_index("datetime")
    .groupby("machineID")["maintenance_event"]
    .rolling("30D", closed="left")
    .sum()
    .reset_index()
    .rename(
        columns={
            "maintenance_event": "maintenance_count"
        }
    )
)


# Most recent maintenance date
last_maintenance = (
    maintenance_history[
        ["machineID", "datetime"]
    ]
    .drop_duplicates()
    .sort_values(["datetime", "machineID"])
)


print("Maintenance events:", len(maintenance_history))
print("30-day maintenance table:", maintenance_30d.shape)
```

```python
# Feature Engineering — Cell 5B
# Calculate days since previous maintenance

df = df.sort_values(
    ["datetime", "machineID"]
).reset_index(drop=True)

last_maintenance = last_maintenance.sort_values(
    ["datetime", "machineID"]
).reset_index(drop=True)

# Find most recent maintenance before current telemetry time
df = pd.merge_asof(
    df,
    last_maintenance.rename(
        columns={"datetime": "last_maintenance_datetime"}
    ),
    left_on="datetime",
    right_on="last_maintenance_datetime",
    by="machineID",
    direction="backward",
    allow_exact_matches=False
)

# Calculate days since maintenance
df["days_since_maintenance"] = (
    df["datetime"] - df["last_maintenance_datetime"]
).dt.total_seconds() / (24 * 60 * 60)

# Merge 30-day maintenance count
maintenance_30d = maintenance_30d.sort_values(
    ["datetime", "machineID"]
).reset_index(drop=True)

df = pd.merge_asof(
    df,
    maintenance_30d,
    on="datetime",
    by="machineID",
    direction="backward",
    allow_exact_matches=False
)

# Machines with no previous maintenance
# Use a large but reasonable value to represent unknown history
df["days_since_maintenance"] = (
    df["days_since_maintenance"].fillna(999)
)

df["maintenance_count"] = (
    df["maintenance_count"].fillna(0)
)

print("Shape:", df.shape)

df[[
    "datetime",
    "machineID",
    "maintenance_count",
    "days_since_maintenance"
]].head(20)
```

```python
# Feature Engineering — Cell 6
# Add machine model and age

machine_info = machines[
    ["machineID", "model", "age"]
].copy()

machine_info = machine_info.rename(
    columns={
        "age": "machine_age_years"
    }
)

df = df.merge(
    machine_info,
    on="machineID",
    how="left"
)

print("Shape:", df.shape)

df[[
    "datetime",
    "machineID",
    "model",
    "machine_age_years"
]].head(20)
```

```python
# Cell 7 — Create time-based features

# Make sure datetime is datetime type
df["datetime"] = pd.to_datetime(df["datetime"])

# Basic time features
df["hour"] = df["datetime"].dt.hour
df["day_of_week"] = df["datetime"].dt.dayofweek
df["month"] = df["datetime"].dt.month

# Weekend indicator
df["is_weekend"] = (df["day_of_week"] >= 5).astype(int)

# Cyclical encoding for hour
df["hour_sin"] = np.sin(2 * np.pi * df["hour"] / 24)
df["hour_cos"] = np.cos(2 * np.pi * df["hour"] / 24)

# Cyclical encoding for day of week
df["dow_sin"] = np.sin(2 * np.pi * df["day_of_week"] / 7)
df["dow_cos"] = np.cos(2 * np.pi * df["day_of_week"] / 7)

print("Shape:", df.shape)

df[
    [
        "datetime",
        "hour",
        "day_of_week",
        "month",
        "is_weekend",
        "hour_sin",
        "hour_cos",
        "dow_sin",
        "dow_cos"
    ]
].head(20)
```

```python
# Cell 8 — Create next-24-hour failure target

# Prepare failure data
failure_events = failures[["machineID", "datetime", "failure"]].copy()

failure_events["datetime"] = pd.to_datetime(failure_events["datetime"])

failure_events = failure_events.sort_values(
    ["datetime", "machineID"]
).reset_index(drop=True)

# Prepare telemetry data
df = df.sort_values(
    ["datetime", "machineID"]
).reset_index(drop=True)

# Find the next failure for each machine
target_data = pd.merge_asof(
    df[["datetime", "machineID"]],
    failure_events,
    on="datetime",
    by="machineID",
    direction="forward",
    allow_exact_matches=False,
    tolerance=pd.Timedelta(hours=24)
)

# Create binary target
df["failure_next_24h"] = target_data["failure"].notna().astype(int)

# Store upcoming component for root-cause analysis
df["next_failure_component"] = target_data["failure"]

# Store upcoming failure time for validation/analysis
df["next_failure_time"] = target_data["datetime"]

print("Shape:", df.shape)

print("\nTarget distribution:")
print(df["failure_next_24h"].value_counts())

print("\nTarget percentage:")
print(
    (df["failure_next_24h"].value_counts(normalize=True) * 100)
    .round(3)
)

print("\nPositive examples:")
display(
    df.loc[df["failure_next_24h"] == 1,
           ["datetime", "machineID", "failure_next_24h",
            "next_failure_component", "next_failure_time"]].head(20)
)
```

```python
# Cell 8B — Correctly store the actual future failure timestamp

# Prepare failures with a separate timestamp column
failure_events = failures[["machineID", "datetime", "failure"]].copy()

failure_events["datetime"] = pd.to_datetime(failure_events["datetime"])
failure_events["failure_time"] = failure_events["datetime"]

failure_events = failure_events[
    ["machineID", "datetime", "failure", "failure_time"]
].sort_values(["datetime", "machineID"]).reset_index(drop=True)

# Re-sort telemetry
df = df.sort_values(
    ["datetime", "machineID"]
).reset_index(drop=True)

# Find next failure within 24 hours
target_data = pd.merge_asof(
    df[["datetime", "machineID"]],
    failure_events,
    on="datetime",
    by="machineID",
    direction="forward",
    allow_exact_matches=False,
    tolerance=pd.Timedelta(hours=24)
)

# Keep the already-created target
df["failure_next_24h"] = target_data["failure"].notna().astype(int)

# Actual future failure information
df["next_failure_component"] = target_data["failure"]
df["next_failure_time"] = target_data["failure_time"]

print("Shape:", df.shape)

print("\nPositive examples:")
display(
    df.loc[
        df["failure_next_24h"] == 1,
        [
            "datetime",
            "machineID",
            "failure_next_24h",
            "next_failure_component",
            "next_failure_time"
        ]
    ].head(20)
)
```

```python
# Cell 9 — Validate next-24-hour target

# Check positive rows
positive_rows = df[df["failure_next_24h"] == 1].copy()

# Calculate hours between current telemetry and future failure
positive_rows["hours_to_failure"] = (
    positive_rows["next_failure_time"] - positive_rows["datetime"]
).dt.total_seconds() / 3600

print("Total positive rows:", len(positive_rows))

print("\nHours to failure:")
print(positive_rows["hours_to_failure"].describe())

print("\nMinimum hours to failure:",
      positive_rows["hours_to_failure"].min())

print("Maximum hours to failure:",
      positive_rows["hours_to_failure"].max())

# Validation checks
print("\nValidation checks:")

print(
    "Any failure at or before current timestamp:",
    (positive_rows["hours_to_failure"] <= 0).any()
)

print(
    "Any failure beyond 24 hours:",
    (positive_rows["hours_to_failure"] > 24).any()
)

print(
    "Missing future failure time among positives:",
    positive_rows["next_failure_time"].isna().sum()
)

# Show a few examples
print("\nSample validated positive rows:")

display(
    positive_rows[
        [
            "datetime",
            "machineID",
            "failure_next_24h",
            "next_failure_component",
            "next_failure_time",
            "hours_to_failure"
        ]
    ].head(20)
)
```

```python
# Cell 10 — Time-based Train / Validation / Test split

# Make sure data is sorted chronologically
df = df.sort_values(
    ["datetime", "machineID"]
).reset_index(drop=True)

# Get unique timestamps
unique_times = df["datetime"].sort_values().unique()

# Calculate split points
train_end = unique_times[int(len(unique_times) * 0.70)]
val_end = unique_times[int(len(unique_times) * 0.85)]

# Create splits
train_df = df[df["datetime"] < train_end].copy()

val_df = df[
    (df["datetime"] >= train_end) &
    (df["datetime"] < val_end)
].copy()

test_df = df[
    df["datetime"] >= val_end
].copy()

print("TRAIN")
print("Shape:", train_df.shape)
print("Start:", train_df["datetime"].min())
print("End:", train_df["datetime"].max())

print("\nVALIDATION")
print("Shape:", val_df.shape)
print("Start:", val_df["datetime"].min())
print("End:", val_df["datetime"].max())

print("\nTEST")
print("Shape:", test_df.shape)
print("Start:", test_df["datetime"].min())
print("End:", test_df["datetime"].max())

print("\nTarget distribution")

print("\nTrain:")
print(train_df["failure_next_24h"].value_counts())
print((train_df["failure_next_24h"].value_counts(normalize=True) * 100).round(3))

print("\nValidation:")
print(val_df["failure_next_24h"].value_counts())
print((val_df["failure_next_24h"].value_counts(normalize=True) * 100).round(3))

print("\nTest:")
print(test_df["failure_next_24h"].value_counts())
print((test_df["failure_next_24h"].value_counts(normalize=True) * 100).round(3))
```

```python
# Cell 11 — Sensor deviation from 24-hour baseline

sensor_columns = ["volt", "rotate", "pressure", "vibration"]

for col in sensor_columns:
    df[f"{col}_deviation_24h"] = (
        df[col] - df[f"{col}_mean_24h"]
    )

print("Shape:", df.shape)

display(
    df[
        [
            "datetime",
            "machineID",
            "volt",
            "volt_deviation_24h",
            "rotate",
            "rotate_deviation_24h",
            "pressure",
            "pressure_deviation_24h",
            "vibration",
            "vibration_deviation_24h"
        ]
    ].head(20)
)
```

```python
# Cell 12 — Final feature audit

print("Shape:", df.shape)

print("\nDuplicate rows:", df.duplicated().sum())

print("\nMissing values:")
missing = df.isna().sum()
display(
    missing[missing > 0]
    .sort_values(ascending=False)
    .to_frame("missing_count")
)

print("\nData types:")
print(df.dtypes.value_counts())

print("\nTarget columns:")
print([
    col for col in df.columns
    if "failure" in col.lower()
])

print("\nAll columns:")
for i, col in enumerate(df.columns, 1):
    print(f"{i:02d}. {col}")
```

```python
# Cell 13 — Prepare final ML feature datasets

# Columns that must NOT be used as model features
drop_columns = [
    "datetime",
    "machineID",
    "last_maintenance_datetime",
    "next_failure_time",
    "next_failure_component",
    "failure_next_24h"
]

# Features
feature_columns = [
    col for col in df.columns
    if col not in drop_columns
]

# Create X and y
X = df[feature_columns].copy()
y = df["failure_next_24h"].copy()

print("Number of features:", len(feature_columns))
print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nFeature columns:")
for i, col in enumerate(feature_columns, 1):
    print(f"{i:02d}. {col}")

print("\nFeature data types:")
print(X.dtypes.value_counts())

print("\nTarget distribution:")
print(y.value_counts())
```

```python
# Cell 14 — Create X/y train, validation and test sets

# Use the datetime boundaries from our existing splits
train_mask = df["datetime"] < train_end

val_mask = (
    (df["datetime"] >= train_end) &
    (df["datetime"] < val_end)
)

test_mask = df["datetime"] >= val_end

# Features
X_train = X.loc[train_mask].copy()
X_val = X.loc[val_mask].copy()
X_test = X.loc[test_mask].copy()

# Target
y_train = y.loc[train_mask].copy()
y_val = y.loc[val_mask].copy()
y_test = y.loc[test_mask].copy()

print("X_train:", X_train.shape)
print("X_val:  ", X_val.shape)
print("X_test: ", X_test.shape)

print("\ny_train:", y_train.shape)
print("y_val:  ", y_val.shape)
print("y_test: ", y_test.shape)

print("\nPositive cases:")
print("Train:", y_train.sum())
print("Validation:", y_val.sum())
print("Test:", y_test.sum())
```

```python
# Cell 15 — Preprocessing pipeline

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

# Identify column types
categorical_features = ["model"]

numeric_features = [
    col for col in X_train.columns
    if col not in categorical_features
]

print("Numerical features:", len(numeric_features))
print("Categorical features:", len(categorical_features))

# Numerical preprocessing
numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median"))
    ]
)

# Categorical preprocessing
categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=True
        ))
    ]
)

# Combined preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

print("\nPreprocessor created successfully.")
```

```python
# Cell 16 — Fit preprocessing and transform datasets

X_train_processed = preprocessor.fit_transform(X_train)

X_val_processed = preprocessor.transform(X_val)

X_test_processed = preprocessor.transform(X_test)

print("Original feature count:", X_train.shape[1])

print("Processed X_train shape:", X_train_processed.shape)
print("Processed X_val shape:  ", X_val_processed.shape)
print("Processed X_test shape: ", X_test_processed.shape)

print("\nProcessed data type:", type(X_train_processed))
```

```python
# Cell 17 — Logistic Regression baseline

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix
)

# Calculate class weight
negative_count = (y_train == 0).sum()
positive_count = (y_train == 1).sum()

scale_pos_weight = negative_count / positive_count

print("Negative cases:", negative_count)
print("Positive cases:", positive_count)
print("Scale positive weight:", scale_pos_weight)

# Logistic Regression
logistic_model = LogisticRegression(
    class_weight="balanced",
    max_iter=1000,
    solver="liblinear",
    random_state=42
)

print("\nTraining Logistic Regression...")

logistic_model.fit(
    X_train_processed,
    y_train
)

print("Training completed.")

# Validation predictions
y_val_pred = logistic_model.predict(X_val_processed)
y_val_prob = logistic_model.predict_proba(X_val_processed)[:, 1]

# Metrics
print("\nValidation Results")
print("------------------")

print("Accuracy :", round(accuracy_score(y_val, y_val_pred), 4))
print("Precision:", round(precision_score(y_val, y_val_pred, zero_division=0), 4))
print("Recall   :", round(recall_score(y_val, y_val_pred, zero_division=0), 4))
print("F1 Score :", round(f1_score(y_val, y_val_pred, zero_division=0), 4))
print("ROC-AUC  :", round(roc_auc_score(y_val, y_val_prob), 4))
print("PR-AUC   :", round(average_precision_score(y_val, y_val_prob), 4))

print("\nConfusion Matrix:")
print(confusion_matrix(y_val, y_val_pred))
```

```python
# Cell 18 — Threshold tuning on validation set

import numpy as np
import pandas as pd
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score
)

threshold_results = []

for threshold in np.arange(0.05, 0.96, 0.05):

    y_threshold_pred = (
        y_val_prob >= threshold
    ).astype(int)

    threshold_results.append({
        "threshold": round(threshold, 2),
        "precision": precision_score(
            y_val,
            y_threshold_pred,
            zero_division=0
        ),
        "recall": recall_score(
            y_val,
            y_threshold_pred,
            zero_division=0
        ),
        "f1": f1_score(
            y_val,
            y_threshold_pred,
            zero_division=0
        )
    })

threshold_df = pd.DataFrame(threshold_results)

print("Threshold performance:")
display(
    threshold_df.round(4)
)

# Best threshold according to F1
best_threshold_row = threshold_df.loc[
    threshold_df["f1"].idxmax()
]

best_threshold = best_threshold_row["threshold"]

print("\nBest threshold based on F1:")
print("Threshold:", best_threshold)
print("Precision:", round(best_threshold_row["precision"], 4))
print("Recall:", round(best_threshold_row["recall"], 4))
print("F1:", round(best_threshold_row["f1"], 4))
```

```python
# Cell 19 — Final Logistic Regression evaluation on TEST

# Generate test probabilities
y_test_prob_logistic = logistic_model.predict_proba(
    X_test_processed
)[:, 1]

# Apply threshold selected from validation
y_test_pred_logistic = (
    y_test_prob_logistic >= best_threshold
).astype(int)

# Calculate metrics
test_accuracy = accuracy_score(
    y_test,
    y_test_pred_logistic
)

test_precision = precision_score(
    y_test,
    y_test_pred_logistic,
    zero_division=0
)

test_recall = recall_score(
    y_test,
    y_test_pred_logistic,
    zero_division=0
)

test_f1 = f1_score(
    y_test,
    y_test_pred_logistic,
    zero_division=0
)

test_roc_auc = roc_auc_score(
    y_test,
    y_test_prob_logistic
)

test_pr_auc = average_precision_score(
    y_test,
    y_test_prob_logistic
)

print("LOGISTIC REGRESSION — TEST RESULTS")
print("-----------------------------------")
print("Threshold:", best_threshold)
print("Accuracy :", round(test_accuracy, 4))
print("Precision:", round(test_precision, 4))
print("Recall   :", round(test_recall, 4))
print("F1 Score :", round(test_f1, 4))
print("ROC-AUC  :", round(test_roc_auc, 4))
print("PR-AUC   :", round(test_pr_auc, 4))

print("\nConfusion Matrix:")
print(
    confusion_matrix(
        y_test,
        y_test_pred_logistic
    )
)
```

```python
# ============================================================
# RANDOM FOREST — TRAINING
# ============================================================

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix
)

rf_model = RandomForestClassifier(
    n_estimators=300,
    max_depth=18,
    min_samples_leaf=2,
    class_weight="balanced_subsample",
    n_jobs=-1,
    random_state=42
)

print("Training Random Forest...")

rf_model.fit(X_train_processed, y_train)

print("Random Forest training completed.")
```

```python
# ============================================================
# STEP 14 — INSPECT MISSING TELEMETRY
# ============================================================

missing_rc = rootcause_data[
    rootcause_data[["volt", "rotate", "pressure", "vibration"]]
    .isna()
    .any(axis=1)
].copy()

print("Number of rows with missing telemetry:", len(missing_rc))

print("\nMissing rows:")
display(
    missing_rc[
        ["machineID", "datetime_x", "target_time", "datetime_y", "failure"]
    ]
)

print("\nFailure components among missing rows:")
print(missing_rc["failure"].value_counts())
```

```python
# ============================================================
# STEP 15 — REMOVE EVENTS WITHOUT 24-HOUR TELEMETRY
# ============================================================

rootcause_model_data = rootcause_data.dropna(
    subset=["volt", "rotate", "pressure", "vibration"]
).copy()

print("Original failure events:", len(rootcause_data))
print("Usable failure events:", len(rootcause_model_data))
print("Excluded events:", len(rootcause_data) - len(rootcause_model_data))

print("\nComponent distribution after exclusion:")
print(
    rootcause_model_data["failure"]
    .value_counts()
    .sort_index()
)
```

```python
# ============================================================
# STEP 16 — BUILD ROOT-CAUSE FEATURE DATASET
# ============================================================

# Merge the 24-hour-before-failure prediction points
# with our already engineered feature dataset.
component_df = rootcause_model_data.merge(
    df,
    left_on=["machineID", "target_time"],
    right_on=["machineID", "datetime"],
    how="inner",
    suffixes=("_failure", "_feature")
)

print("Component feature dataset shape:", component_df.shape)

print("\nComponent distribution:")
print(component_df["failure"].value_counts().sort_index())
```

```python
# ============================================================
# STEP 16B — DEFINE TARGET AND REMOVE LEAKAGE COLUMNS
# ============================================================

target_col = "failure"

leakage_columns = [
    "failure",
    "datetime_failure",
    "datetime_feature",
    "datetime",
    "target_time",
    "machineID",
    "failure_next_24h",
    "next_failure_component",
    "next_failure_time",
    "last_maintenance_datetime"
]

leakage_columns = [
    col for col in leakage_columns
    if col in component_df.columns
]

X_component = component_df.drop(
    columns=leakage_columns,
    errors="ignore"
)

y_component = component_df[target_col]

print("X_component shape:", X_component.shape)
print("y_component shape:", y_component.shape)

print("\nFeature columns:")
print(X_component.columns.tolist())

print("\nTarget distribution:")
print(y_component.value_counts())
```

```python
# ============================================================
# STEP 18 — CHRONOLOGICAL TRAIN / VALIDATION / TEST SPLIT
# ============================================================

# Sort by the ACTUAL failure timestamp
component_sorted = component_df.sort_values(
    "datetime_x"
).reset_index(drop=True)

n = len(component_sorted)

train_end = int(n * 0.70)
val_end = int(n * 0.85)

component_train = component_sorted.iloc[:train_end].copy()
component_val = component_sorted.iloc[train_end:val_end].copy()
component_test = component_sorted.iloc[val_end:].copy()

print("Dataset sizes:")
print("Train:", len(component_train))
print("Validation:", len(component_val))
print("Test:", len(component_test))

print("\nDate ranges:")

print(
    "Train:",
    component_train["datetime_x"].min(),
    "→",
    component_train["datetime_x"].max()
)

print(
    "Validation:",
    component_val["datetime_x"].min(),
    "→",
    component_val["datetime_x"].max()
)

print(
    "Test:",
    component_test["datetime_x"].min(),
    "→",
    component_test["datetime_x"].max()
)

print("\nComponent distribution:")

print("\nTRAIN")
print(component_train["failure"].value_counts().sort_index())

print("\nVALIDATION")
print(component_val["failure"].value_counts().sort_index())

print("\nTEST")
print(component_test["failure"].value_counts().sort_index())
```

```python
# ============================================================
# STEP 19 — CREATE TRAIN / VALIDATION / TEST FEATURES
# CORRECTED FOR MERGED COLUMN NAMES
# ============================================================

# Mapping from original feature names to their names in component_df
column_mapping = {
    "volt": "volt_feature",
    "rotate": "rotate_feature",
    "pressure": "pressure_feature",
    "vibration": "vibration_feature",
    "month": "month_feature"
}


def build_component_features(data):
    """
    Build the 50-feature matrix using the correct column
    names from component_df.
    """

    result = pd.DataFrame(index=data.index)

    for feature in component_feature_columns:

        source_column = column_mapping.get(
            feature,
            feature
        )

        if source_column not in data.columns:
            raise KeyError(
                f"Required feature '{feature}' "
                f"could not be found as '{source_column}'"
            )

        result[feature] = data[source_column]

    return result


# Build feature matrices
X_component_train = build_component_features(component_train)
X_component_val = build_component_features(component_val)
X_component_test = build_component_features(component_test)


# Targets
y_component_train = component_train["failure"].copy()
y_component_val = component_val["failure"].copy()
y_component_test = component_test["failure"].copy()


# Check results
print("TRAIN:")
print("X:", X_component_train.shape)
print("y:", y_component_train.shape)

print("\nVALIDATION:")
print("X:", X_component_val.shape)
print("y:", y_component_val.shape)

print("\nTEST:")
print("X:", X_component_test.shape)
print("y:", y_component_test.shape)

print("\nCategorical features:")
print(
    X_component_train
    .select_dtypes(include=["object"])
    .columns.tolist()
)

print("\nNumeric features:")
print(
    len(
        X_component_train
        .select_dtypes(exclude=["object"])
        .columns
    )
)
```

```python
# ============================================================
# STEP 20 — COMPONENT MODEL PREPROCESSING
# ============================================================

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

# Identify feature types
component_categorical_features = ["model"]

component_numeric_features = [
    col for col in X_component_train.columns
    if col not in component_categorical_features
]

# Numeric preprocessing
component_numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median"))
])

# Categorical preprocessing
component_categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=True
    ))
])

# Combined preprocessing pipeline
component_preprocessor = ColumnTransformer([
    (
        "num",
        component_numeric_transformer,
        component_numeric_features
    ),
    (
        "cat",
        component_categorical_transformer,
        component_categorical_features
    )
])

# FIT ONLY ON TRAINING DATA
X_component_train_processed = component_preprocessor.fit_transform(
    X_component_train
)

# Transform validation and test
X_component_val_processed = component_preprocessor.transform(
    X_component_val
)

X_component_test_processed = component_preprocessor.transform(
    X_component_test
)

print("Processed TRAIN:", X_component_train_processed.shape)
print("Processed VALIDATION:", X_component_val_processed.shape)
print("Processed TEST:", X_component_test_processed.shape)
```

```python
# ============================================================
# STEP 21 — ENCODE COMPONENT TARGET LABELS
# ============================================================

from sklearn.preprocessing import LabelEncoder

component_label_encoder = LabelEncoder()

# Fit ONLY on training labels
y_component_train_encoded = component_label_encoder.fit_transform(
    y_component_train
)

# Transform validation and test
y_component_val_encoded = component_label_encoder.transform(
    y_component_val
)

y_component_test_encoded = component_label_encoder.transform(
    y_component_test
)

print("Component mapping:")

for i, label in enumerate(component_label_encoder.classes_):
    print(f"{label} → {i}")

print("\nEncoded TRAIN distribution:")
print(
    pd.Series(y_component_train_encoded)
    .value_counts()
    .sort_index()
)

print("\nEncoded VALIDATION distribution:")
print(
    pd.Series(y_component_val_encoded)
    .value_counts()
    .sort_index()
)

print("\nEncoded TEST distribution:")
print(
    pd.Series(y_component_test_encoded)
    .value_counts()
    .sort_index()
)
```

```python
# ============================================================
# STEP 22 — TRAIN MULTICLASS XGBOOST ROOT-CAUSE MODEL
# ============================================================

from xgboost import XGBClassifier

component_xgb_model = XGBClassifier(
    objective="multi:softprob",
    num_class=4,
    n_estimators=300,
    learning_rate=0.05,
    max_depth=5,
    min_child_weight=2,
    subsample=0.8,
    colsample_bytree=0.8,
    eval_metric="mlogloss",
    tree_method="hist",
    random_state=42,
    n_jobs=-1
)

component_xgb_model.fit(
    X_component_train_processed,
    y_component_train_encoded,
    eval_set=[
        (
            X_component_val_processed,
            y_component_val_encoded
        )
    ],
    verbose=50
)

print("\nComponent XGBoost training completed.")
```

```python
# ============================================================
# STEP 23 — VALIDATION EVALUATION
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# Predict class probabilities
val_proba = component_xgb_model.predict_proba(
    X_component_val_processed
)

# Convert probabilities to predicted class
val_pred = np.argmax(val_proba, axis=1)

# Overall metrics
val_accuracy = accuracy_score(
    y_component_val_encoded,
    val_pred
)

val_precision = precision_score(
    y_component_val_encoded,
    val_pred,
    average="macro",
    zero_division=0
)

val_recall = recall_score(
    y_component_val_encoded,
    val_pred,
    average="macro",
    zero_division=0
)

val_f1 = f1_score(
    y_component_val_encoded,
    val_pred,
    average="macro",
    zero_division=0
)

print("VALIDATION RESULTS")
print("=" * 50)

print(f"Accuracy:          {val_accuracy:.4f}")
print(f"Macro Precision:   {val_precision:.4f}")
print(f"Macro Recall:      {val_recall:.4f}")
print(f"Macro F1:          {val_f1:.4f}")

print("\nClassification Report:")
print(
    classification_report(
        y_component_val_encoded,
        val_pred,
        target_names=component_label_encoder.classes_,
        zero_division=0
    )
)

print("\nConfusion Matrix:")
print(
    confusion_matrix(
        y_component_val_encoded,
        val_pred
    )
)
```

```python
# ============================================================
# STEP 24 — FIND EXACT BEST VALIDATION ITERATION
# ============================================================

eval_history = component_xgb_model.evals_result()

val_logloss = eval_history["validation_0"]["mlogloss"]

best_iteration = int(np.argmin(val_logloss))
best_logloss = val_logloss[best_iteration]

print("Best iteration:", best_iteration)
print("Best validation mlogloss:", best_logloss)

print("\nNearby validation losses:")

start = max(0, best_iteration - 5)
end = min(len(val_logloss), best_iteration + 6)

for i in range(start, end):
    print(f"Iteration {i}: {val_logloss[i]:.6f}")
```

```python
# ============================================================
# STEP 25 — FINAL ROOT-CAUSE XGBOOST MODEL
# ============================================================

component_xgb_final = XGBClassifier(
    objective="multi:softprob",
    num_class=4,
    n_estimators=131,
    learning_rate=0.05,
    max_depth=5,
    min_child_weight=2,
    subsample=0.8,
    colsample_bytree=0.8,
    eval_metric="mlogloss",
    tree_method="hist",
    random_state=42,
    n_jobs=-1
)

component_xgb_final.fit(
    X_component_train_processed,
    y_component_train_encoded,
    verbose=False
)

print("Final component model trained.")
print("Number of trees:", component_xgb_final.n_estimators)
```

```python
# ============================================================
# STEP 26 — FINAL TEST EVALUATION
# ============================================================

# Predict probabilities
test_proba = component_xgb_final.predict_proba(
    X_component_test_processed
)

# Predicted class
test_pred = np.argmax(test_proba, axis=1)

# Metrics
test_accuracy = accuracy_score(
    y_component_test_encoded,
    test_pred
)

test_precision = precision_score(
    y_component_test_encoded,
    test_pred,
    average="macro",
    zero_division=0
)

test_recall = recall_score(
    y_component_test_encoded,
    test_pred,
    average="macro",
    zero_division=0
)

test_f1 = f1_score(
    y_component_test_encoded,
    test_pred,
    average="macro",
    zero_division=0
)

print("FINAL TEST RESULTS")
print("=" * 50)

print(f"Accuracy:          {test_accuracy:.4f}")
print(f"Macro Precision:   {test_precision:.4f}")
print(f"Macro Recall:      {test_recall:.4f}")
print(f"Macro F1:          {test_f1:.4f}")

print("\nClassification Report:")
print(
    classification_report(
        y_component_test_encoded,
        test_pred,
        target_names=component_label_encoder.classes_,
        zero_division=0
    )
)

print("\nConfusion Matrix:")
print(
    confusion_matrix(
        y_component_test_encoded,
        test_pred
    )
)
```

```python
# ============================================================
# STEP 31 — LSTM DATA PREPARATION
# ============================================================

# Make sure telemetry is sorted correctly
telemetry_lstm = telemetry.copy()

telemetry_lstm["datetime"] = pd.to_datetime(
    telemetry_lstm["datetime"]
)

telemetry_lstm = telemetry_lstm.sort_values(
    ["machineID", "datetime"]
).reset_index(drop=True)

# Core telemetry features
lstm_features = [
    "volt",
    "rotate",
    "pressure",
    "vibration"
]

# Use the same chronological boundaries as the main ML model
train_start = df["datetime"].min()
train_end = df.iloc[613199]["datetime"]

val_start = df.iloc[613200]["datetime"]
val_end = df.iloc[744599]["datetime"]

test_start = df.iloc[744600]["datetime"]
test_end = df["datetime"].max()

print("LSTM FEATURES:")
print(lstm_features)

print("\nTIME SPLIT")
print("=" * 50)
print("Train:", train_start, "→", train_end)
print("Validation:", val_start, "→", val_end)
print("Test:", test_start, "→", test_end)

print("\nTelemetry shape:", telemetry_lstm.shape)
```

```python
# ============================================================
# STEP 32 — SCALE LSTM TELEMETRY FEATURES
# ============================================================

from sklearn.preprocessing import StandardScaler

lstm_scaler = StandardScaler()

# Training telemetry only
train_mask = (
    (telemetry_lstm["datetime"] >= train_start) &
    (telemetry_lstm["datetime"] <= train_end)
)

val_mask = (
    (telemetry_lstm["datetime"] >= val_start) &
    (telemetry_lstm["datetime"] <= val_end)
)

test_mask = (
    (telemetry_lstm["datetime"] >= test_start) &
    (telemetry_lstm["datetime"] <= test_end)
)

# Fit scaler ONLY on training data
lstm_scaler.fit(
    telemetry_lstm.loc[train_mask, lstm_features]
)

# Transform all periods
telemetry_lstm_scaled = telemetry_lstm.copy()

telemetry_lstm_scaled[lstm_features] = (
    lstm_scaler.transform(
        telemetry_lstm[lstm_features]
    )
)

print("Scaler fitted on training data only.")

print("\nTraining scaled means:")
print(
    telemetry_lstm_scaled.loc[
        train_mask, lstm_features
    ].mean().round(4)
)

print("\nTraining scaled standard deviations:")
print(
    telemetry_lstm_scaled.loc[
        train_mask, lstm_features
    ].std().round(4)
)
```

```python
# ============================================================
# STEP 33 — CREATE LSTM TARGET
# ============================================================

# Create failure target lookup from the engineered dataset
lstm_target_lookup = df[
    ["machineID", "datetime", "failure_next_24h"]
].copy()

# Merge target with scaled telemetry
lstm_data = telemetry_lstm_scaled.merge(
    lstm_target_lookup,
    on=["machineID", "datetime"],
    how="left"
)

# Safety check
lstm_data["failure_next_24h"] = (
    lstm_data["failure_next_24h"]
    .fillna(0)
    .astype("int8")
)

# Sort again
lstm_data = lstm_data.sort_values(
    ["machineID", "datetime"]
).reset_index(drop=True)

print("LSTM dataset shape:", lstm_data.shape)

print("\nTarget distribution:")
print(
    lstm_data["failure_next_24h"]
    .value_counts()
    .sort_index()
)

print("\nTarget rate:")
print(
    lstm_data["failure_next_24h"].mean()
)
```

```python
# ============================================================
# STEP 34 — CHECK HOURLY SEQUENCE CONTINUITY
# ============================================================

# Calculate time difference between consecutive records
lstm_data["time_diff"] = (
    lstm_data.groupby("machineID")["datetime"]
    .diff()
)

# Count gaps
gap_counts = lstm_data["time_diff"].value_counts()

print("Most common time intervals:")
print(gap_counts.head(10))

print("\nNumber of non-hourly gaps:")
non_hourly_gaps = (
    (lstm_data["time_diff"].notna()) &
    (lstm_data["time_diff"] != pd.Timedelta(hours=1))
).sum()

print(non_hourly_gaps)

print("\nMaximum gap:")
print(lstm_data["time_diff"].max())
```

```python
# ============================================================
# STEP 35 — CREATE 24-HOUR LSTM SEQUENCES
# ============================================================

import numpy as np

sequence_length = 24

sequence_X = []
sequence_y = []
sequence_times = []

# Remove helper column before sequence creation
lstm_data = lstm_data.drop(columns=["time_diff"])

for machine_id, machine_data in lstm_data.groupby("machineID"):

    machine_data = machine_data.sort_values("datetime")

    sensor_values = machine_data[lstm_features].values.astype(
        np.float32
    )

    targets = machine_data["failure_next_24h"].values.astype(
        np.int8
    )

    timestamps = machine_data["datetime"].values

    # Create sequences
    for i in range(sequence_length, len(machine_data)):

        sequence_X.append(
            sensor_values[i-sequence_length:i]
        )

        sequence_y.append(
            targets[i]
        )

        sequence_times.append(
            timestamps[i]
        )

# Convert to NumPy arrays
sequence_X = np.asarray(sequence_X, dtype=np.float32)
sequence_y = np.asarray(sequence_y, dtype=np.int8)
sequence_times = np.asarray(sequence_times)

print("Sequence X shape:", sequence_X.shape)
print("Sequence y shape:", sequence_y.shape)
print("Sequence timestamps:", sequence_times.shape)

print("\nTarget distribution:")
print(
    pd.Series(sequence_y)
    .value_counts()
    .sort_index()
)
```

```python
# ============================================================
# STEP 36 — CHRONOLOGICAL LSTM TRAIN / VAL / TEST SPLIT
# ============================================================

train_seq_mask = sequence_times <= np.datetime64(train_end)

val_seq_mask = (
    (sequence_times >= np.datetime64(val_start)) &
    (sequence_times <= np.datetime64(val_end))
)

test_seq_mask = (
    (sequence_times >= np.datetime64(test_start)) &
    (sequence_times <= np.datetime64(test_end))
)

X_lstm_train = sequence_X[train_seq_mask]
y_lstm_train = sequence_y[train_seq_mask]

X_lstm_val = sequence_X[val_seq_mask]
y_lstm_val = sequence_y[val_seq_mask]

X_lstm_test = sequence_X[test_seq_mask]
y_lstm_test = sequence_y[test_seq_mask]

print("LSTM SPLIT")
print("=" * 50)

print("Train X:", X_lstm_train.shape)
print("Train y:", y_lstm_train.shape)

print("Validation X:", X_lstm_val.shape)
print("Validation y:", y_lstm_val.shape)

print("Test X:", X_lstm_test.shape)
print("Test y:", y_lstm_test.shape)

print("\nPositive counts:")
print("Train:", y_lstm_train.sum())
print("Validation:", y_lstm_val.sum())
print("Test:", y_lstm_test.sum())

print("\nPositive rates:")
print("Train:", y_lstm_train.mean())
print("Validation:", y_lstm_val.mean())
print("Test:", y_lstm_test.mean())
```

```python
# ============================================================
# STEP 37 — LSTM CLASS WEIGHT
# ============================================================

negative_count = (y_lstm_train == 0).sum()
positive_count = (y_lstm_train == 1).sum()

lstm_class_weight = {
    0: 1.0,
    1: negative_count / positive_count
}

print("LSTM CLASS WEIGHT")
print("=" * 50)

print("Negative samples:", negative_count)
print("Positive samples:", positive_count)
print(
    "Positive class weight:",
    round(lstm_class_weight[1], 4)
)
```

```python
# ============================================================
# STEP 38 — BUILD LSTM MODEL
# ============================================================

from tensorflow import keras
from tensorflow.keras import layers

# Clear any previous Keras model from memory
keras.backend.clear_session()

lstm_model = keras.Sequential([
    layers.Input(shape=(24, 4)),

    layers.LSTM(
        64,
        return_sequences=False
    ),

    layers.Dropout(0.30),

    layers.Dense(
        32,
        activation="relu"
    ),

    layers.Dropout(0.20),

    layers.Dense(
        1,
        activation="sigmoid"
    )
])

lstm_model.compile(
    optimizer=keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="binary_crossentropy",
    metrics=[
        keras.metrics.Precision(name="precision"),
        keras.metrics.Recall(name="recall"),
        keras.metrics.AUC(
            name="roc_auc",
            curve="ROC"
        ),
        keras.metrics.AUC(
            name="pr_auc",
            curve="PR"
        )
    ]
)

lstm_model.summary()
```

```python
# ============================================================
# STEP 39 — TRAIN LSTM
# ============================================================

early_stopping = keras.callbacks.EarlyStopping(
    monitor="val_pr_auc",
    mode="max",
    patience=3,
    restore_best_weights=True,
    verbose=1
)

reduce_lr = keras.callbacks.ReduceLROnPlateau(
    monitor="val_pr_auc",
    mode="max",
    factor=0.5,
    patience=1,
    min_lr=1e-6,
    verbose=1
)

history = lstm_model.fit(
    X_lstm_train,
    y_lstm_train,

    validation_data=(
        X_lstm_val,
        y_lstm_val
    ),

    epochs=15,
    batch_size=256,

    class_weight=lstm_class_weight,

    callbacks=[
        early_stopping,
        reduce_lr
    ],

    verbose=1
)
```

```python
# ============================================================
# STEP 40 — LSTM TEST EVALUATION
# ============================================================

# Predict failure probabilities
lstm_test_proba = lstm_model.predict(
    X_lstm_test,
    batch_size=256,
    verbose=1
).ravel()

# Default threshold
lstm_test_pred = (
    lstm_test_proba >= 0.5
).astype(int)

# Metrics
lstm_accuracy = accuracy_score(
    y_lstm_test,
    lstm_test_pred
)

lstm_precision = precision_score(
    y_lstm_test,
    lstm_test_pred,
    zero_division=0
)

lstm_recall = recall_score(
    y_lstm_test,
    lstm_test_pred,
    zero_division=0
)

lstm_f1 = f1_score(
    y_lstm_test,
    lstm_test_pred,
    zero_division=0
)

lstm_roc_auc = roc_auc_score(
    y_lstm_test,
    lstm_test_proba
)

lstm_pr_auc = average_precision_score(
    y_lstm_test,
    lstm_test_proba
)

print("\nLSTM FINAL TEST RESULTS")
print("=" * 50)

print(f"Accuracy:    {lstm_accuracy:.4f}")
print(f"Precision:   {lstm_precision:.4f}")
print(f"Recall:      {lstm_recall:.4f}")
print(f"F1 Score:    {lstm_f1:.4f}")
print(f"ROC-AUC:     {lstm_roc_auc:.4f}")
print(f"PR-AUC:      {lstm_pr_auc:.4f}")

print("\nConfusion Matrix:")
print(
    confusion_matrix(
        y_lstm_test,
        lstm_test_pred
    )
)

print("\nClassification Report:")
print(
    classification_report(
        y_lstm_test,
        lstm_test_pred,
        zero_division=0
    )
)
```

```python
# ============================================================
# STEP 41 — LSTM VALIDATION THRESHOLD TUNING
# ============================================================

# Get validation probabilities
lstm_val_proba = lstm_model.predict(
    X_lstm_val,
    batch_size=256,
    verbose=1
).ravel()

threshold_results_lstm = []

for threshold in np.arange(0.05, 0.96, 0.05):

    val_pred_threshold = (
        lstm_val_proba >= threshold
    ).astype(int)

    precision = precision_score(
        y_lstm_val,
        val_pred_threshold,
        zero_division=0
    )

    recall = recall_score(
        y_lstm_val,
        val_pred_threshold,
        zero_division=0
    )

    f1 = f1_score(
        y_lstm_val,
        val_pred_threshold,
        zero_division=0
    )

    threshold_results_lstm.append({
        "threshold": round(threshold, 2),
        "precision": precision,
        "recall": recall,
        "f1": f1
    })

lstm_threshold_df = pd.DataFrame(
    threshold_results_lstm
)

best_lstm_threshold_row = (
    lstm_threshold_df
    .loc[lstm_threshold_df["f1"].idxmax()]
)

print("BEST LSTM VALIDATION THRESHOLD")
print("=" * 50)

print(
    best_lstm_threshold_row
)

print("\nAll threshold results:")
display(lstm_threshold_df)
```

```python
# ============================================================
# STEP 45F — REBUILD CLEAN SCORING DATASET
# ============================================================

# Start again from the original test rows
final_scoring = df.iloc[
    test_start_index:test_end_index
].copy()

# Add XGBoost failure prediction
final_scoring["failure_probability"] = xgb_test_proba

final_scoring["predicted_failure"] = (
    xgb_test_proba >= xgb_threshold
).astype(int)

# Application risk level
final_scoring["risk_level"] = pd.cut(
    final_scoring["failure_probability"],
    bins=[-np.inf, 0.30, 0.70, np.inf],
    labels=["Low", "Medium", "High"]
)

# Merge unique component predictions
final_scoring = final_scoring.merge(
    component_unique,
    left_on=["machineID", "datetime"],
    right_on=["machineID", "target_time"],
    how="left"
)

# Remove helper column
final_scoring = final_scoring.drop(
    columns=["target_time"],
    errors="ignore"
)

print("FINAL SCORING DATASET")
print("=" * 50)

print("Shape:", final_scoring.shape)

print(
    "Expected rows:",
    test_end_index - test_start_index
)

print(
    "Actual rows:",
    len(final_scoring)
)

print(
    "\nRows with component prediction:",
    final_scoring["predicted_component"].notna().sum()
)

print("\nRisk distribution:")
print(
    final_scoring["risk_level"]
    .value_counts()
    .sort_index()
)

print("\nComponent distribution:")
print(
    final_scoring["predicted_component"]
    .value_counts(dropna=True)
)
```

```python
# ============================================================
# STEP 47 — GENERATE TOP SHAP FACTORS
# ============================================================

# Feature names from the fitted preprocessor
shap_feature_names = preprocessor.get_feature_names_out()

# Function to get top positive SHAP factors
def get_top_shap_factors(shap_row, feature_names, top_n=3):
    positive_indices = np.argsort(shap_row)[::-1]

    factors = []

    for idx in positive_indices:
        if shap_row[idx] > 0:
            clean_name = feature_names[idx].replace(
                "num__", ""
            ).replace(
                "cat__model_",
                "model_"
            )

            factors.append(clean_name)

            if len(factors) == top_n:
                break

    return factors


# Generate top 3 positive factors for every test row
top_shap_factors = []

for row in shap_test_values:
    top_shap_factors.append(
        get_top_shap_factors(
            row,
            shap_feature_names,
            top_n=3
        )
    )

# Convert to columns
final_scoring["top_factor_1"] = [
    factors[0] if len(factors) > 0 else None
    for factors in top_shap_factors
]

final_scoring["top_factor_2"] = [
    factors[1] if len(factors) > 1 else None
    for factors in top_shap_factors
]

final_scoring["top_factor_3"] = [
    factors[2] if len(factors) > 2 else None
    for factors in top_shap_factors
]

print("SHAP factors added successfully.")

display(
    final_scoring[
        [
            "machineID",
            "datetime",
            "failure_probability",
            "risk_level",
            "top_factor_1",
            "top_factor_2",
            "top_factor_3"
        ]
    ].head(15)
)
```

```python
# ============================================================
# STEP 49 — MAINTENANCE RECOMMENDATIONS
# ============================================================

def generate_recommendation(row):

    risk = row["risk_level"]
    component = row["predicted_component"]

    if risk == "High":

        if component == "comp1":
            return "Urgent inspection of Component 1. Schedule preventive maintenance immediately."

        elif component == "comp2":
            return "Urgent inspection of Component 2. Schedule preventive maintenance immediately."

        elif component == "comp3":
            return "Urgent inspection of Component 3. Schedule preventive maintenance immediately."

        elif component == "comp4":
            return "Urgent inspection of Component 4. Schedule preventive maintenance immediately."

        else:
            return "High failure risk. Schedule immediate preventive inspection."

    elif risk == "Medium":

        if component in ["comp1", "comp2", "comp3", "comp4"]:
            return f"Monitor machine closely and inspect {component} during the next maintenance cycle."

        else:
            return "Increase monitoring frequency and review maintenance history."

    else:

        return "Normal operation. Continue routine monitoring."


final_scoring["maintenance_recommendation"] = (
    final_scoring.apply(
        generate_recommendation,
        axis=1
    )
)

print("Maintenance recommendations added.")

display(
    final_scoring[
        [
            "machineID",
            "datetime",
            "failure_probability",
            "risk_level",
            "predicted_component",
            "component_confidence",
            "maintenance_recommendation"
        ]
    ]
    .dropna(subset=["predicted_component"])
    .head(15)
)
```
